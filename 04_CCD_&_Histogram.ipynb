```python
import numpy as np
import scipy.stats as st
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x = st.norm.rvs(loc=0, scale=1, size=10000)
plt.hist(x,bins=50,range=(-5,5),density=True) #blue histogram
xp = np.linspace(-5,5,50)
plt.plot(xp,st.norm.pdf(xp,loc=0,scale=1),lw=2) #orange line, normal pdf
plt.show()
```

```python
# 1. Generate or load your data
# (Replacing this with your own array or dataframe column)
np.random.seed(0)
data = np.random.normal(loc=50, scale=5, size=1000)

# 2. Fit a normal distribution (mean and standard deviation) to the data
mu,std = st.norm.fit(data)
print(f"Fitted Mean (mu): {mu:.2f}")
print(f"Fitted Standard Deviation (std): {std:.2f}")

# 3. Plot the histogram of the data
plt.hist(data, bins=30, density=True, alpha=0.6, color='r', label='Data Histogram')

# 4. Plot the best-fit normal distribution curve
xmin, xmax = plt.xlim()
x = np.linspace(xmin, xmax, 100)
p = st.norm.pdf(x, mu, std)
plt.plot(x, p, 'k', linewidth=2, label='Fitted Normal PDF')

plt.title(f"Fit results: mu = {mu:.2f},  std = {std:.2f}")
plt.legend()
plt.show()
```
